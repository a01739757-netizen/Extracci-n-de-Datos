```python
#Importar librerias
import pandas as pd

```

```python
#Carga archivo
df = pd.read_excel('Actividad 1.1 (Citas_Digital)/Citas_Digital.xlsx')
df
```

```python
#Verificar info
df.info()
```

```python
filtro1= df.iloc[0:673,: ]
filtro1.info()

#Filtro por filas
```

```python
#filtros columnas
filtro2 = filtro1.iloc[ : , [0,1,2,4,5,6,7,8,9]].copy() # Forma equivalente
filtro2.info()
```

```python
#Convertimos la tabña a texto para poder limpiarla por valores numéricos y con errores de captura
filtro2['Potencial de compra'] = filtro2['Potencial de compra'].astype(str)

```

```python
#Reemplazo de datos de una columna
filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("Ventas ","0", regex=False)


filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("%","", regex=False)


filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("'","", regex=False)


filtro2['Potencial de compra']=filtro2['Potencial de compra'].str.replace("&","", regex=False)


#Conversión de  columna a tipo String a Float
filtro2['Potencial de compra']= filtro2['Potencial de compra'].astype(float)
filtro2.info()
```

```python
#Convertir a CSV
filtro2.to_csv("Citas_Digital_General.csv", index=False)

```
